## **Logistic Regression using perceptron trick**

```python
def perceptron(x, y):

    x = np.insert(x, 0, 1, axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(0,100)
        y_pred = step(np.dot(x[j], weights))
        weights = weights + lr*(y[j] - y_pred)*x[j]

    return weights[0], weights[1:]

# this step funtion checks whether the prediction lies in positive area or negative
def step(number):
    return 1 if number>0 else 0
```

```python
intercept_,coef_ = perceptron(x,y)
```

coef is coefficient of A and B

intercept is C

```python
print(coef_)
print(intercept_)
```

from equation Ax + By + C = 0

value of m = - A / B

value of b = - C / B

to convert Ax + By + C = 0 into the y = mx + b

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_ / coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m * x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input, y_input, color = 'red', linewidth = 3)
plt.scatter(x[:,0], x[:,1], c=y, cmap='winter', s=100) #c gives the color to each point based on the value of y
plt.ylim(-3, 2)
```

## **Logistic regression with sigmoid function**

```python
def perceptron(x,y):
    
    x = np.insert(x,0,1,axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.1
    
    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = sigmoid(np.dot(x[j],weights))
        weights = weights + lr*(y[j]-y_hat)*x[j]
        
    return weights[0],weights[1:]

def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
intercept_,coef_ = perceptron(x,y)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input3 = np.linspace(-3,3,100)
y_input3 = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.plot(x_input2,y_input2,color='black',linewidth=3)
plt.plot(x_input3,y_input3,color='brown',linewidth=3)
plt.scatter(x[:,0],x[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
